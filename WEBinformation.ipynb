```python
documents = {
"D1": "Information retrieval helps users find relevant documents.",
"D2": "Information Retrieval systems index documents and documents.",
"D3": "Search engines retrieve relevant web documents.",
"D4": "Document indexing supports efficient information retrieval.",
"D5": "Machine learning improves search engine ranking.",
"D6": "Search systems use indexed documents to retrieve information.",
"D7": "The search engine retrieves documents; the documents are searchable.",
"D8": "Indexing documents makes information retrieval more efficient.",
"D9": "DOCUMENTS contain information about document indexing.",
"D10": "Retrieval systems organise documents and document collections."
}

import re
def preprocess_document(text):
   text = text.lower()
   text = re.sub(r'[^\w\s]', '', text)
   tokens = text.split()

   return tokens



A = {}    # dictionnaire pour stocker les résultats
# Appliquer le preprocessing à chaque document
for doc, text in documents.items():
    A[doc] = preprocess_document(text)


# Afficher les mots de chaque document
for doc, tokens in A.items():
    print(doc, ":", tokens)
```

```python
import nltk


nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))

B = {}
for doc, tokens in A.items():# Parcourir tous les documents


    B[doc] = [w for w in tokens if w not in stop_words]  # Garder les mots qui ne sont pas des stop words


for doc, tokens in B.items():
    print(doc, ":", tokens)  #  les résultats
```

```python
# Importer PorterStemmer
from nltk.stem import PorterStemmer

# Créer un stemmer
stemmer = PorterStemmer()

# Dictionnaire pour stocker les résultats
C = {}

# Appliquer le stemming aux mots de chaque document
for doc, tokens in B.items():
    C[doc] = [stemmer.stem(w) for w in tokens]

# Afficher les résultats
for doc, tokens in C.items():
    print(doc, ":", tokens)
```

```python
# Comparer les trois versions du corpus
for doc in documents:
    print("\n", doc)
    print("Version A :", A[doc])
    print("Version B :", B[doc])
    print("Version C :", C[doc])
```

```python
# Construire le vocabulaire de chaque version

vocabulary_A = sorted(set(w for tokens in A.values() for w in tokens))
vocabulary_B = sorted(set(w for tokens in B.values() for w in tokens))
vocabulary_C = sorted(set(w for tokens in C.values() for w in tokens))

# Afficher les vocabulaires
print("Vocabulary A:", vocabulary_A)
print("Vocabulary B:", vocabulary_B)
print("Vocabulary C:", vocabulary_C)
```

```python

def build_term_document_matrix(tokenized_documents):


    vocabulary = sorted(set(w for tokens in tokenized_documents.values() for w in tokens))


    matrix = {}


    for term in vocabulary:
        matrix[term] = {}


        for doc, tokens in tokenized_documents.items():
            if term in tokens:
                matrix[term][doc] = 1
            else:
                matrix[term][doc] = 0

    return matrix



binary_matrix = build_term_document_matrix(C)


for term, values in binary_matrix.items():
    print(term, values)
```

```python

def build_tf_matrix(tokenized_documents):


    vocabulary = sorted(set(w for tokens in tokenized_documents.values() for w in tokens))


    matrix = {}


    for term in vocabulary:
        matrix[term] = {}


        for doc, tokens in tokenized_documents.items():
            matrix[term][doc] = tokens.count(term)

    return matrix



tf_matrix = build_tf_matrix(C)


for term, values in tf_matrix.items():
    print(term, values)
```

```python
import pandas as pd

# Afficher la matrice binaire
print("Binary Matrix:")
df_binary = pd.DataFrame(binary_matrix).T
print(df_binary)

# Afficher la matrice TF
print("\nTerm Frequency Matrix:")
df_tf = pd.DataFrame(tf_matrix).T
print(df_tf)
```

```python

versions = {
    "A": A,
    "B": B,
    "C": C
}


results = {}


for name, collection in versions.items():


    vocabulary = set(w for tokens in collection.values() for w in tokens)
    vocabulary_size = len(vocabulary)


    total_tokens = sum(len(tokens) for tokens in collection.values())


    average_tokens = total_tokens / len(collection)


    total_postings = 0

    for term in vocabulary:
        for tokens in collection.values():
            if term in tokens:
                total_postings += 1


    token_positions = total_tokens


    results[name] = [
        vocabulary_size,
        average_tokens,
        total_postings,
        token_positions
    ]


for name, values in results.items():
    print("\nVersion", name)
    print("Vocabulary size:", values[0])
    print("Average tokens:", round(values[1], 2))
    print("Total postings:", values[2])
    print("Token positions:", values[3])
```

```python
import pandas as pd

# Créer le tableau de comparaison
table = pd.DataFrame(
    results,
    index=[
        "Vocabulary size",
        "Average tokens per document",
        "Total postings",
        "Token positions"
    ]
)


print(table)
```

```python

def search_term(term, collection):

    result = []


    for doc, tokens in collection.items():


        if term in tokens:
            result.append(doc)

    return result



print(search_term("information", A))
```

```python

def build_inverted_index(collection):

    index = {}


    for doc, tokens in collection.items():


        for i, word in enumerate(tokens):


            if word not in index:
                index[word] = {}


            if doc not in index[word]:
                index[word][doc] = []

            index[word][doc].append(i)

    return index



inverted_index = build_inverted_index(C)


print(inverted_index)
```

```python

manual_index = {
    "inform": {"D1": [0], "D2": [0]},
    "retriev": {"D1": [1], "D2": [1], "D3": [2]},
    "help": {"D1": [2]},
    "user": {"D1": [3]},
    "find": {"D1": [4]},
    "relev": {"D1": [5], "D3": [3]},
    "document": {"D1": [6], "D2": [4, 5], "D3": [5]},
    "system": {"D2": [2]},
    "index": {"D2": [3]},
    "search": {"D3": [0]},
    "engin": {"D3": [1]},
    "web": {"D3": [4]}
}


# Garder uniquement D1, D2 et D3
automatic_index = {}

for word, docs in inverted_index.items():
    for doc, positions in docs.items():

        if doc in ["D1", "D2", "D3"]:

            if word not in automatic_index:
                automatic_index[word] = {}

            automatic_index[word][doc] = positions

# Comparer les deux index
print("Result:", automatic_index == manual_index)
```

```python

binary_matrix = build_term_document_matrix(C)
tf_matrix = build_tf_matrix(C)



for term in binary_matrix:
    for doc in binary_matrix[term]:


        if binary_matrix[term][doc] not in [0, 1]:
            print("Binary matrix error")

print("Binary matrix is correct!")



for term in tf_matrix:
    for doc in tf_matrix[term]:


        if tf_matrix[term][doc] != C[doc].count(term):
            print("TF matrix error")

print("TF matrix is correct!")
```
